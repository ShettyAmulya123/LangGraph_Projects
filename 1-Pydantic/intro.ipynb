```python
from pydantic import BaseModel
```

```python
class Person(BaseModel):
    name:str
    age:int
    city:str

person=Person(name="Amulya",age=20,city="Mangalore") 

print(person)
```

```python
from dataclasses import dataclass

@dataclass
class Person():
    name:str
    age:int
    city:str

person=Person(name="Amulya",age=20,city="Mangalore") 

print(person)
```
